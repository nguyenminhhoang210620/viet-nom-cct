```python
from vocab import Vocab, PAD, UNK

vi_tokens = ["xin", "chào", "bạn", PAD]       # sample Vietnamese tokens
no_tokens = ["漢", "喃", UNK]                 # sample Hán/Nôm tokens (include [UNK] to test filtering)
```

```python
v = Vocab.build(vi_tokens, no_tokens)
```

```python
v
```

```python
print("id of 'chào':", v.encode("chào"))
print("id of unknown token:", v.encode("n/a"))      # returns UNK_ID (1)
print("decode id 2:", v.decode(2))
```

```python


print("size:", len(v))
print("id2token:", v.id2token)
print("token2id:", v.token2id)

# encode / decode
print("id of 'chào':", v.encode("chào"))
print("id of unknown token:", v.encode("n/a"))      # returns UNK_ID (1)
print("decode id 2:", v.decode(2))

# sequence encoding
seq_ids = v.encode_seq(["xin", "bạn", "???"])
print("seq ids:", seq_ids)

# partitions
print("n_vi, n_no:", v.n_vi, v.n_no)
print("is id 2 a vi id?", v.is_vi_id(2))
print("target ids:", v.target_vocab_ids())

# save & load
v.save("vocab.json")
v2 = Vocab.load("vocab.json")
print("loaded equals original:", v2.id2token == v.id2token)
```
